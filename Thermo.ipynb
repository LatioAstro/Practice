```python

def multiplicity_dipole(up,down):
    total = up+down
    
def entropy(omega):
    return const.k_B * np.log(omega)

def v_rms(mass,temperature):
    return np.sqrt(3*const.k_B*temperature/mass)

def sackur_tetrode(N,V,U,m):
    """
    N: Number of particles
    V: Volume of the monatomic ideal gas
    U: Internal energy of the gas
    m: total mass of the gas
    """
    inner = V/N * ( (4 * np.pi * m * U) / (3*N*const.h**2) ) ** (3/2)
    return N * const.k_B *( np.log(inner) + (5/2) )

def enthapy(U,P,V):
    return U + P*V

def helmholtz_free_energy(U,T,S):
    return U - T*S

def gibbs_free_energy(F,P,V):
    return F + P*V

def Clausius_Clapeyron(L,T,V1,V2):
    return L/(T*(V2-V1))
```
